```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("tennis.csv")
df
```

```python
y = df["play"]
x = df.drop("play", axis = 1)
```

```python
length_y = len(y)
p_y = y.value_counts() / length_y
y_entropy = -(p_y * np.log2(p_y)).sum()
y_entropy
```

```python
def y_inside_each_group_temp(y, x = None):
  a = x.unique()
  add = []
  print(a)
  for i in a:
    length_i = (x == i).sum()
    p_i = y[x == i].value_counts() / length_i
    i_entropy = -(p_i * np.log2(p_i)).sum()
    print(f"{i} : {i_entropy}")
    freq_i = length_i / len(x)
    formula = freq_i * i_entropy
    add.append(formula)
  return sum(add)
y_inside_each_group_temp(y, x["temp"])
```

```python
def y_inside_each_group_humidity(y, x = None):
  a = x.unique()
  add = []
  print(a)
  for i in a:
    length_i = (x == i).sum()
    p_i = y[x == i].value_counts() / length_i
    i_entropy = -(p_i * np.log2(p_i)).sum()
    print(f"{i} : {i_entropy}")
    freq_i = length_i / len(x)
    formula = freq_i * i_entropy
    add.append(formula)
  return sum(add)
y_inside_each_group_humidity(y, x["humidity"])
```

```python
def y_inside_each_group_outlook(y, x = None):
  a = x.unique()
  add = []
  print(a)
  for i in a:
    length_i = (x == i).sum()
    p_i = y[x == i].value_counts() / length_i
    i_entropy = -(p_i * np.log2(p_i)).sum()
    print(f"{i} : {i_entropy}")
    freq_i = length_i / len(x)
    formula = freq_i * i_entropy
    add.append(formula)
  return sum(add)
y_inside_each_group_outlook(y, x["outlook"])
```

```python
def y_inside_each_group_windy(y, x = None):
  a = x.unique()
  add = []
  print(a)
  for i in a:
    length_i = (x == i).sum()
    p_i = y[x == i].value_counts() / length_i
    i_entropy = -(p_i * np.log2(p_i)).sum()
    print(f"{i} : {i_entropy}")
    freq_i = length_i / len(x)
    formula = freq_i * i_entropy
    add.append(formula)
  return sum(add)
y_inside_each_group_windy(y, x["windy"])
```

```python
print("TEMP:---------")
print(f"info_gain: {y_entropy - y_inside_each_group_temp(y, x["temp"])}")
print("HUMIDITY:---------")
print(f"info_gain: {y_entropy - y_inside_each_group_humidity(y, x["humidity"])}")
print("OUTLOOK:---------")
print(f"info_gain: {y_entropy - y_inside_each_group_outlook(y, x["outlook"])}")
print("WINDY:---------")
print(f"info_gain: {y_entropy - y_inside_each_group_windy(y, x["windy"])}")
```
