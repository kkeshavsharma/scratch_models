```python
df = pd.read_csv("tennis.csv")
df
```

```python
y = df["play"]
x = df.drop("play", axis = 1)
```

```python
length_y = len(y)
p_y = y.value_counts() / length_y
y_entropy = -(p_y * np.log2(p_y)).sum()
y_entropy
```

```python
arr = [x["temp"], x["outlook"], x["humidity"], x["windy"]]
def entropy(arr):
  for i in arr:
    length_i = len(i)
    p_i = i.value_counts() / length_i
    i_entropy = -(p_i * np.log2(p_i)).sum()
    print(f"{i.name} : {i_entropy}")
entropy(arr)
```
